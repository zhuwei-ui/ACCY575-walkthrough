## 3. What's weird?

```python
for col in ["entity_type", "audit_type", "gaap_results", "is_low_risk_auditee"]:
    print(f"--- {col} ---")
    print(df[col].value_counts(dropna=False).head(10), "\n")
```

```python
# placeholder strings masquerading as data — isna() will not catch these
for col in ["auditor_certify_name", "auditor_certify_title"]:
    print(f"--- {col} ---")
    print(df[col].value_counts(dropna=False).head(5), "\n")
```

```python
# empty strings used instead of nulls?
obj_cols = df.select_dtypes(include=["object", "str"]).columns
blank = {c: int((df[c].fillna("").astype(str).str.strip() == "").sum()) for c in obj_cols}
pd.Series(blank).sort_values(ascending=False).head(15)
```

```python
# do any fiscal years run backwards?
start = pd.to_datetime(df["fy_start_date"], errors="coerce")
end = pd.to_datetime(df["fy_end_date"], errors="coerce")

print(f"unparseable fy_start_date : {int(start.isna().sum()):,}")
print(f"unparseable fy_end_date   : {int(end.isna().sum()):,}")
print(f"end before start          : {int((end < start).sum()):,}")
print(f"\nfy length (days) summary:")
(end - start).dt.days.describe()
```

```python
amt = pd.to_numeric(df["total_amount_expended"], errors="coerce")
print(f"non-numeric : {int((amt.isna() & df['total_amount_expended'].notna()).sum()):,}")
print(f"zero        : {int((amt == 0).sum()):,}")
print(f"negative    : {int((amt < 0).sum()):,}")
amt.describe()
```

```python
print(f"duplicate report_id : {int(df['report_id'].duplicated().sum()):,}")
print(f"duplicate full rows : {int(df.duplicated().sum()):,}")
```
